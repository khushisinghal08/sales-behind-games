```python
import numpy as np
import pandas as pd
import ast
```

```python
url = "https://raw.githubusercontent.com/khushisinghal08/sales-behind-games/main/games_file.csv"

df = pd.read_csv(url)

df.head()
```

```python
df.info()
```

```python
# Normalizing the Column Names
df.columns = df.columns.str.strip().str.title()

df=df.rename(columns={'Unnamed: 0':'Game_Id','Release Date':'Release_Date','Times Listed':'Times_Listed','Number Of Reviews':'Number_Of_Reviews'})
#df.head(5)
df['Release_Date'].apply(type).unique()
df.info()
```

```python
# Changing Data Types
df['Release_Date'] = pd.to_datetime(df['Release_Date'], dayfirst=True,format='mixed',errors='coerce')

# Changing Data Types and Replacing 'K','M' Values from Dataset
columns=['Times_Listed','Number_Of_Reviews','Plays','Playing','Backlogs','Wishlist']
for col in columns:
    df[col]=df[col].str.replace('k','e3',case=False).str.replace('m','e6', case=False).astype(float)



# df.head(10)
```

```python
# Replacing Empty Strings from Team and Genres
cols=['Genres','Team']
for col in cols:
    df[col] = df[col].apply(lambda x: ast.literal_eval(x) if isinstance(x,str) else x)
    df[col] = df[col].apply(lambda x: np.nan if x == [] else x)


# Handling NA Values
df['Team']=df['Team'].fillna('Unknown')
df['Genres']=df['Genres'].fillna('Unknown')
df['Summary']=df['Summary'].fillna('Not Given')
for col in cols:
    df = df.explode(col)

# Normalizing the Genres and Team Values

for col in cols :
    df[col]=df[col].astype(str).str.strip().str.title()
# df['Rating'].unique() # <-- Checking If any Value is an Outlier or not.
df['Rating']=df['Rating'].fillna(df.groupby('Genres')['Rating'].transform('mean'))

# df[df['Release_Date'].isna()] # <-- Checking How many Null Values Records.
df=df.dropna(subset=['Release_Date'])


#df.isna().mean()
df.duplicated().sum()
```
